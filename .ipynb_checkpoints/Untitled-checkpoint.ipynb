```python
import pandas as pd
from statsmodels.tsa.statespace.sarimax import SARIMAX
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# ---------------------------
# 1. Prepare Data
# ---------------------------
def prepare_data(train, test, feature_columns, target_column):
    """
    Combine training and test data for SARIMAX.
    """
    # Sort by player and season
    train = train.sort_values(by=['name', 'season']).reset_index(drop=True)
    test = test.sort_values(by=['name', 'season']).reset_index(drop=True)
    
    # Combine train and test for continuity
    train['is_test'] = 0
    test['is_test'] = 1
    combined = pd.concat([train, test], ignore_index=True)

    # Filter relevant columns
    combined = combined[['name', 'season'] + feature_columns + [target_column, 'is_test']]
    return combined

# Specify columns for each timestep
feature_columns = ['swstr%', 'o-swing%', 'z-swing%', 'swing%', 
                   'o-contact%', 'z-contact%', 'contact%', 'zone%', 'cstr%', 'age', 'stuff+','pitching+','location+']
target_column = 'k%'

# Prepare combined data
combined_data = prepare_data(train, test, feature_columns, target_column)

# ---------------------------
# 2. Train SARIMAX (2019–2023 Data for 2024 Prediction)
# ---------------------------
def train_sarimax(data, feature_columns, target_column, sarimax_order=(1, 0, 0)):
    """
    Train a single SARIMAX model using all players' data.
    """
    # Use only training data (2019–2023)
    train_data = data[data['is_test'] == 0]
    
    # Define endog (target) and exog (features)
    endog = train_data[target_column]  # Target variable: k%
    exog = train_data[feature_columns]  # Exogenous predictors
    
    # Train SARIMAX
    model = SARIMAX(endog, exog=exog, order=sarimax_order)
    results = model.fit(disp=False)
    return results

# Train the SARIMAX model
sarimax_model = train_sarimax(combined_data, feature_columns, target_column)

# ---------------------------
# 3. Predict 2024 K% (Test Set)
# ---------------------------
def predict_sarimax(model, data, feature_columns, target_column):
    """
    Predict K% for the test set (2024).
    """
    # Use test data (2024) for prediction
    test_data = data[data['is_test'] == 1]
    
    # Define exog (features for test set)
    exog = test_data[feature_columns]
    
    # Predict for 2024
    predictions = model.forecast(steps=len(test_data), exog=exog)
    
    # Combine predictions with test data for evaluation
    test_data = test_data.copy()
    test_data['k%_pred'] = predictions
    return test_data

# Predict K% for 2024
predictions_2024 = predict_sarimax(sarimax_model, combined_data, feature_columns, target_column)

# ---------------------------
# 4. Predict 2025 K% (Future Step)
# ---------------------------
def predict_2025(model, data, feature_columns):
    """
    Extend predictions to 2025 using 2024 data as input.
    """
    # Use 2024 data as input for 2025 predictions
    test_data = data[data['season'] == 2024]
    
    # Define exog (features for 2024)
    exog = test_data[feature_columns]
    
    # Predict for 2025
    predictions = model.forecast(steps=len(test_data), exog=exog)
    
    # Combine predictions with a placeholder for 2025
    results_2025 = test_data.copy()
    results_2025['season'] = 2025
    results_2025['k%_pred'] = predictions
    return results_2025

# Predict K% for 2025
predictions_2025 = predict_2025(sarimax_model, combined_data, feature_columns)

# ---------------------------
# 5. Evaluate and Save Results
# ---------------------------
# Evaluate 2024 Predictions
y_test_2024 = predictions_2024[target_column]
y_pred_2024 = predictions_2024['k%_pred']

mae = mean_absolute_error(y_test_2024, y_pred_2024)
mse = mean_squared_error(y_test_2024, y_pred_2024)
rmse = np.sqrt(mse)
r2 = r2_score(y_test_2024, y_pred_2024)

print(f"SARIMAX Metrics for 2024:\nMAE: {mae}\nMSE: {mse}\nRMSE: {rmse}\nR2: {r2}")

# Save predictions to CSV
predictions_2024.to_csv('sarimax_predictions_2024.csv', index=False)
predictions_2025.to_csv('sarimax_predictions_2025.csv', index=False)

print("Predictions saved to sarimax_predictions_2024.csv and sarimax_predictions_2025.csv")
```
